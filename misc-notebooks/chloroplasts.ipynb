```r
# Load packages
suppressPackageStartupMessages({
    library(phyloseq)
    library(tidyverse)
    library(Biostrings)
    library(dada2)
})
    
options(repr.plot.width=15, repr.plot.height=8)
```

```r
# Read in filtered raw abundance phyloseq object
ps_rel <- readRDS("/scratch4/workspace/jade_fiorilla_student_uml_edu-rawseqorcc/dada2/ps-objects/ps-dada2-rel.rds")
```

```r
# Keep only chloroplasts
ps_chlor <- subset_taxa(ps_rel, Order == "Chloroplast")
```

```r
# Check chloroplast taxonomic assignments
tax_table(ps_chlor) %>%
  as.data.frame() %>%
  head(10)
```

Seems like SILVA was unable to classify the chloroplasts beyond the order level - can try a different reference database.

```r
# Extract sequences
chlor_seqs <- refseq(ps_chlor)

# Create named DNAStringSet
chlor_fasta <- DNAStringSet(as.character(chlor_seqs))
names(chlor_fasta) <- taxa_names(ps_chlor)
writeXStringSet(chlor_fasta, "/scratch4/workspace/jade_fiorilla_student_uml_edu-rawseqorcc/dada2/chloroplast_ASVs.fasta")
```

```r
# Assign taxonomy using dada2 and PhytoREF database
chlor_tax <- assignTaxonomy(chlor_fasta, "/work/pi_sarah_gignouxwolfsohn_uml_edu/jade/pr2_version_5.1.1_SSU_dada2.fasta.gz", multithread = TRUE, minBoot = 50)
```
